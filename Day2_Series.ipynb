# Pandas Series Exercises

This section covers answers and practical examples for the questions provided.

### Q1. Create pandas series from different data types like list, numpy array and dictionary.

```python
import numpy as np
import pandas as pd

# 1. From a List
list_data = ['Apple', 'Banana', 'Cherry']
series_list = pd.Series(list_data)
print("Series from List:")
display(series_list)

# 2. From a NumPy Array
array_data = np.array([10, 20, 30, 40])
series_array = pd.Series(array_data, index=['a', 'b', 'c', 'd'])
print("\nSeries from NumPy Array:")
display(series_array)

# 3. From a Dictionary
dict_data = {'France': 'Paris', 'Germany': 'Berlin', 'Italy': 'Rome'}
series_dict = pd.Series(dict_data)
print("\nSeries from Dictionary:")
display(series_dict)
```

### Q2. How to get the items of Series A not present in Series B?

```python
series_a = pd.Series([1, 2, 3, 4, 5])
series_b = pd.Series([4, 5, 6, 7, 8])

# Get items in series_a not present in series_b
not_present = series_a[~series_a.isin(series_b)]
print("Items in Series A not in Series B:")
display(not_present)
```

### Q3. How to get intersection of two series?

```python
# We can use numpy.intersect1d or find items present in both
intersection = pd.Series(np.intersect1d(series_a, series_b))
print("Intersection of Series A and Series B:")
display(intersection)
```

### Q4. How to get union of two series?

```python
# Get union of all unique elements from both series
union = pd.Series(np.union1d(series_a, series_b))
print("Union of Series A and Series B:")
display(union)
```

```python
# Simplest forms of operations on two Series
series_a = pd.Series([1, 2, 3, 4, 5])
series_b = pd.Series([4, 5, 6, 7, 8])

# Q2: Items in A not in B
diff = series_a[~series_a.isin(series_b)]
print("A not in B:\n", diff.tolist())

# Q3: Intersection (common items)
inter = np.intersect1d(series_a, series_b)
print("Intersection:\n", inter.tolist())

# Q4: Union (all unique items combined)
union = np.union1d(series_a, series_b)
print("Union:\n", union.tolist())
```
